# Business Insights

### Customer concentration

The analysis identified 4,339 customers with available CustomerID values. The top 10 customers contributed approximately 17.26% of customer-attributed recorded revenue, while the top 20% of identified customers contributed approximately 74.59%.

### High-value customers

Customer 14646 generated approximately €280,206 in recorded revenue across 74 orders and purchased a broad range of products. The customer also placed large recurring orders throughout much of the dataset period, suggesting a bulk or wholesale-style purchasing pattern. This is an analytical inference rather than an explicit customer classification in the dataset.

### Order value

Some countries showed very high average order values because of a small number of unusually large orders. For example, the Netherlands had an average order value of approximately €3,004.70 but a median order value of approximately €788.16. Australia showed a similar pattern, with a mean of approximately €2,430.20 and a median of approximately €429.60. This demonstrates why median order value is useful alongside the mean.

### Monthly performance

Recorded revenue and order volume increased substantially during September–November 2011. November recorded approximately €1.46 million in transaction-line revenue and 3,462 unique invoices. The increase was associated primarily with higher order volume rather than a dramatic change in average order value.

December 2011 is only a partial month because the dataset ends on December 9, so it should not be directly compared with complete months.

### Product performance

Initial product rankings were affected by non-product transaction descriptions such as postage, fees, manual adjustments, and discounts. These were separated from product analysis to produce a more meaningful product-level view.

### Cancellations and reversals

The analysis revealed several products with substantial negative transaction values. Some apparent high-revenue products were strongly affected by cancellations or reversals.

For example, PAPER CRAFT, LITTLE BIRDIE had approximately €168,469.60 in gross positive transaction value but approximately €0 net product revenue after the corresponding negative transaction.

MEDIUM CERAMIC TOP STORAGE JAR also showed a large difference between gross and net revenue, with approximately €81,700.92 gross revenue and €4,221.28 net revenue.

### Data quality lesson

Extreme values should not automatically be removed. They should first be investigated in their business context. In this dataset, extremely large quantities were sometimes associated with exact reversal transactions, meaning that simply treating them as ordinary sales would have produced misleading conclusions.

### Overall conclusion

The investigation demonstrated that meaningful retail analysis requires more than calculating totals. Transaction cancellations, adjustments, missing customer identifiers, extreme orders, non-product transaction lines, and partial periods can substantially affect business metrics. The analysis therefore combined data cleaning, feature engineering, aggregation, visualization, and investigation of unusual records before drawing conclusions.
